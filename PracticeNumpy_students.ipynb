## NumPy Vectorization in practice

Vectorizing your code with NumPy can lead to performance improvements, especially when dealing with large datasets.

The concept of vectorized operations allows the use of more optimal functions and mathematical operations on NumPy array objects.

Here is an example to see the benefits of NumPy vectorization over pure Python, run the following cells and notice the execution time:

```python
# Generate large arrays
size = 1000000
A = np.random.rand(size)
B = np.random.rand(size)
```

```python
# Vectorized approach using NumPy
def vectorized_approach(A, B):
    return A * B
```

```python
# Non-vectorized approach using pure Python
def pure_python_approach(A, B):
    result = []
    for i in range(len(A)):
        result.append(A[i] * B[i])
    return result
```

```python
import time
# Function to calculate execution time
def calculate_execution_time(func, *args):
    start_time = time.time()
    result = func(*args)
    end_time = time.time()
    execution_time = end_time - start_time
    return execution_time, result
```

```python
# Calculate execution time for vectorized approach
vectorized_time, vectorized_result = calculate_execution_time(vectorized_approach, A, B)
print("Vectorized approach execution time:", vectorized_time)

# Calculate execution time for pure Python approach
pure_python_time, pure_python_result = calculate_execution_time(pure_python_approach, A, B)
print("Pure Python approach execution time:", pure_python_time)

print("NumPy is" , round(pure_python_time//vectorized_time), "times faster than pure Python",)

# Verify that the results are the same
print(np.allclose(vectorized_result, pure_python_result))
```
